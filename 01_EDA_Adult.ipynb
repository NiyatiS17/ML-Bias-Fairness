```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Adult dataset directly from URL — no AIF360 needed
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race',
    'sex', 'capital-gain', 'capital-loss', 'hours-per-week',
    'native-country', 'income'
]

df = pd.read_csv(url, names=columns, 
                  na_values=' ?', 
                  skipinitialspace=True)

print("Dataset loaded successfully")
print("Shape:", df.shape)
df.head()
```

```python
# ── CELL 2: Basic Info and Missing Values ─────────
print("=== DATASET SHAPE ===")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n=== MISSING VALUES ===")
print(df.isnull().sum())

print("\n=== DATA TYPES ===")
print(df.dtypes)
```

```python
# ── CELL 3: Target Variable Distribution ──────────
print("=== INCOME DISTRIBUTION ===")
print(df['income'].value_counts())
print("\nPercentage:")
print(df['income'].value_counts(normalize=True).mul(100).round(2))

plt.figure(figsize=(6,4))
df['income'].value_counts().plot(kind='bar', color=['#1B4F8C','#1D6E5E'])
plt.title('Target Variable Distribution — Adult Dataset')
plt.xlabel('Income Category')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ── CELL 4: First Bias Signal — Income by Gender ──
print("=== INCOME >50K BY GENDER ===")
gender_income = df.groupby('sex')['income'].apply(
    lambda x: (x=='>50K').mean()*100).round(2)
print(gender_income)

plt.figure(figsize=(6,4))
gender_income.plot(kind='bar', color=['#1B4F8C','#E67E22'])
plt.title('Percentage Earning >$50K by Gender')
plt.ylabel('Percentage (%)')
plt.xlabel('Gender')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ── CELL 5: Income by Race ────────────────────────
print("=== INCOME >50K BY RACE ===")
race_income = df.groupby('race')['income'].apply(
    lambda x: (x=='>50K').mean()*100).round(2)
print(race_income)

plt.figure(figsize=(9,4))
race_income.plot(kind='bar', color='#1B4F8C')
plt.title('Percentage Earning >$50K by Race')
plt.ylabel('Percentage (%)')
plt.xlabel('Race')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ── CELL 6: Save EDA Summary ──────────────────────
import os
os.makedirs('../results', exist_ok=True)

# Recreate and save gender chart
gender_income = df.groupby('sex')['income'].apply(
    lambda x: (x=='>50K').mean()*100).round(2)

plt.figure(figsize=(6,4))
gender_income.plot(kind='bar', color=['#1B4F8C','#E67E22'])
plt.title('Percentage Earning >$50K by Gender — Adult Dataset')
plt.ylabel('Percentage (%)')
plt.xlabel('Gender')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../results/adult_income_by_gender.png', dpi=150)
plt.show()

# Recreate and save race chart
race_income = df.groupby('race')['income'].apply(
    lambda x: (x=='>50K').mean()*100).round(2)

plt.figure(figsize=(9,4))
race_income.plot(kind='bar', color='#1B4F8C')
plt.title('Percentage Earning >$50K by Race — Adult Dataset')
plt.ylabel('Percentage (%)')
plt.xlabel('Race')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../results/adult_income_by_race.png', dpi=150)
plt.show()

print("Charts saved to results folder")

# Print EDA Summary
print("\n=== EDA SUMMARY — ADULT DATASET ===")
print(f"Total records:         {len(df):,}")
print(f"Missing values:        0")
print(f"Class imbalance:       <=50K: 75.9%  |  >50K: 24.1%")
print(f"Male >50K rate:        {gender_income['Male']:.1f}%")
print(f"Female >50K rate:      {gender_income['Female']:.1f}%")
print(f"Gender gap:            {gender_income['Male']-gender_income['Female']:.1f} percentage points")
print(f"Highest race rate:     Asian-Pac-Islander {race_income['Asian-Pac-Islander']:.1f}%")
print(f"Lowest race rate:      Other {race_income['Other']:.1f}%")
print(f"Racial gap:            {race_income.max()-race_income.min():.1f} percentage points")
print("\nEDA COMPLETE — Adult dataset ready for preprocessing")
```
